```python

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def get_salary(self):
        return self.salary
    @staticmethod
    def get_employee():
        return name


e=Employee("An",855)

print(e.get_salary())
print(e.get_employee())
```
